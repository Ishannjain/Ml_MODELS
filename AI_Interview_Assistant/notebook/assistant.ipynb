## Phase-2 Resume Processing

````python
# ==========================================================
# PHASE 2 : RESUME PROCESSING
# Find Resume Files
# ==========================================================

from pathlib import Path

DATASET_PATH = Path("../dataset")

resume_files = []

for extension in ["*.pdf", "*.docx", "*.txt"]:
    resume_files.extend(DATASET_PATH.glob(extension))

print("=" * 60)
print("RESUME FILES")
print("=" * 60)

if len(resume_files) == 0:
    print("No resume found.")
    print("Place a PDF/DOCX/TXT resume inside ../dataset/")
else:
    for i, file in enumerate(resume_files, 1):
        print(f"{i}. {file.name}")

print("\nTotal Resumes :", len(resume_files))
````

````python
# ==========================================================
# Extract Text From PDF
# ==========================================================

import fitz

pdf_files = list(DATASET_PATH.glob("*.pdf"))

if len(pdf_files) == 0:

    print("No PDF resume found.")

else:

    resume_path = pdf_files[0]

    document = fitz.open(resume_path)

    pages = []

    for page_number, page in enumerate(document):

        text = page.get_text()

        pages.append({
            "page": page_number + 1,
            "text": text
        })

    document.close()

    print("=" * 60)
    print("RESUME TEXT EXTRACTION")
    print("=" * 60)

    print("File :", resume_path.name)
    print("Pages :", len(pages))
````

````python
# ==========================================================
# Combine Extracted Text
# ==========================================================

resume_text = "\n".join(
    page["text"]
    for page in pages
)

print("=" * 60)
print("EXTRACTED RESUME TEXT")
print("=" * 60)

print(resume_text[:3000])

print("\nTotal Characters :", len(resume_text))
print("Total Words :", len(resume_text.split()))
````

````python
# ==========================================================
# Resume Text Cleaning
# ==========================================================

import re

def clean_resume_text(text):

    text = str(text)

    # Normalize line breaks
    text = text.replace("\r", "\n")

    # Remove excessive spaces
    text = re.sub(r"[ \t]+", " ", text)

    # Remove excessive blank lines
    text = re.sub(r"\n{3,}", "\n\n", text)

    return text.strip()


cleaned_resume = clean_resume_text(resume_text)

print("=" * 60)
print("RESUME CLEANING COMPLETED")
print("=" * 60)

print(cleaned_resume[:3000])

print("\nCharacters :", len(cleaned_resume))
print("Words :", len(cleaned_resume.split()))
````

````python
# ==========================================================
# Basic Information Extraction
# ==========================================================

# Email
email_pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}"

emails = re.findall(
    email_pattern,
    cleaned_resume
)

# Phone
phone_pattern = r"(?:\+91[\s-]?)?[6-9]\d{9}"

phones = re.findall(
    phone_pattern,
    cleaned_resume
)

print("=" * 60)
print("BASIC INFORMATION")
print("=" * 60)

print("Email(s) :", list(set(emails)))
print("Phone(s) :", list(set(phones)))
````

````python
# ==========================================================
# Resume Section Detection
# ==========================================================

section_keywords = {

    "education": [
        "education",
        "academic background",
        "qualifications"
    ],

    "experience": [
        "experience",
        "work experience",
        "employment"
    ],

    "skills": [
        "skills",
        "technical skills",
        "core skills"
    ],

    "projects": [
        "projects",
        "academic projects",
        "personal projects"
    ],

    "certifications": [
        "certifications",
        "certificates"
    ],

    "achievements": [
        "achievements",
        "awards"
    ]
}


detected_sections = []

lower_text = cleaned_resume.lower()

for section, keywords in section_keywords.items():

    for keyword in keywords:

        if keyword in lower_text:

            detected_sections.append(section)

            break


detected_sections = list(set(detected_sections))

print("=" * 60)
print("DETECTED RESUME SECTIONS")
print("=" * 60)

for section in detected_sections:

    print("✓", section.title())
````

````python
# ==========================================================
# Skill Extraction
# ==========================================================

skills_database = [

    "Python",
    "Java",
    "C++",
    "C",
    "JavaScript",
    "HTML",
    "CSS",

    "React",
    "Node.js",
    "Django",
    "Flask",

    "SQL",
    "MongoDB",
    "MySQL",

    "Machine Learning",
    "Deep Learning",
    "NLP",
    "Artificial Intelligence",

    "TensorFlow",
    "PyTorch",
    "Scikit-learn",

    "Pandas",
    "NumPy",

    "Git",
    "GitHub",

    "Docker",
    "AWS",

    "Power BI",
    "Tableau"
]


found_skills = []

for skill in skills_database:

    if skill.lower() in lower_text:

        found_skills.append(skill)


print("=" * 60)
print("EXTRACTED SKILLS")
print("=" * 60)

for skill in found_skills:

    print("✓", skill)

print("\nTotal Skills :", len(found_skills))
````

````python
# ==========================================================
# DETERMINE QUESTION CATEGORIES
# ==========================================================

def determine_categories(interview_type):

    if interview_type == "Technical":

        return [
            "Technical",
            "Problem Solving"
        ]

    elif interview_type == "HR":

        return [
            "HR",
            "Behavioral"
        ]

    elif interview_type == "Behavioral":

        return [
            "Behavioral",
            "Situational"
        ]

    elif interview_type == "Mixed":

        return [
            "Technical",
            "Problem Solving",
            "Behavioral",
            "HR"
        ]

    return []


question_categories = determine_categories(
    selected_interview_type
)


print("=" * 60)
print("QUESTION CATEGORIES")
print("=" * 60)

for category in question_categories:

    print("✓", category)
````

````python
# ==========================================================
# QUESTION DISTRIBUTION
# ==========================================================

def create_distribution(total_questions, categories):

    distribution = {}

    category_count = len(categories)

    base_count = total_questions // category_count

    remainder = total_questions % category_count

    for i, category in enumerate(categories):

        count = base_count

        if i < remainder:
            count += 1

        distribution[category] = count

    return distribution


question_distribution = create_distribution(
    selected_question_count,
    question_categories
)


print("=" * 60)
print("QUESTION DISTRIBUTION")
print("=" * 60)

for category, count in question_distribution.items():

    print(f"{category:<20}: {count}")
````

````python
# ==========================================================
# PHASE 4 : QUESTION GENERATION
# ==========================================================

import json
import random

with open(
    "../reports/interview_config.json",
    "r",
    encoding="utf-8"
) as file:

    interview_config = json.load(file)


selected_role = interview_config["job_role"]
selected_experience = interview_config["experience_level"]
selected_interview_type = interview_config["interview_type"]
selected_difficulty = interview_config["difficulty"]
selected_question_count = interview_config["question_count"]

candidate_skills = interview_config["candidate_skills"]
question_distribution = interview_config["question_distribution"]


print("=" * 60)
print("INTERVIEW CONFIGURATION LOADED")
print("=" * 60)

print("Role       :", selected_role)
print("Experience :", selected_experience)
print("Type       :", selected_interview_type)
print("Difficulty :", selected_difficulty)
print("Questions  :", selected_question_count)

print("\nSkills:")
print(candidate_skills)
````

````python
# ==========================================================
# GENERAL TECHNICAL QUESTIONS
# ==========================================================

general_technical_questions = [

    "Explain one technical project you have worked on.",

    "What was the most difficult technical problem you encountered in a project?",

    "How do you debug a program when you encounter an unexpected error?",

    "How do you decide which technology or framework to use for a project?",

    "Explain a technical concept that you recently learned.",

    "How do you improve the performance of a software application?",

    "How do you test your code before deploying an application?"

]
````

````python
# ==========================================================
# RESUME-BASED TECHNICAL QUESTIONS
# ==========================================================

def generate_technical_questions(skills, required_count):

    questions = []

    # First use questions related to resume skills
    for skill in skills:

        if skill in technical_questions:

            available = technical_questions[skill]

            shuffled = available.copy()

            random.shuffle(shuffled)

            for question in shuffled:

                if len(questions) >= required_count:
                    break

                questions.append({

                    "question": question,

                    "category": "Technical",

                    "skill": skill

                })

        if len(questions) >= required_count:
            break


    # Add general questions if needed
    if len(questions) < required_count:

        fallback = general_technical_questions.copy()

        random.shuffle(fallback)

        for question in fallback:

            if len(questions) >= required_count:
                break

            questions.append({

                "question": question,

                "category": "Technical",

                "skill": "General"

            })


    return questions[:required_count]


technical_count = question_distribution.get(
    "Technical",
    0
)

technical_generated = generate_technical_questions(
    candidate_skills,
    technical_count
)


print("=" * 60)
print("TECHNICAL QUESTIONS GENERATED")
print("=" * 60)

for i, item in enumerate(technical_generated, 1):

    print(f"\n{i}. {item['question']}")
    print("Skill :", item["skill"])
````

````python
# ==========================================================
# SITUATIONAL QUESTIONS
# ==========================================================

situational_questions = [

    "If your model performs well during training but poorly on unseen data, what would you do?",

    "If a teammate disagrees with your technical approach, how would you handle the situation?",

    "If you have a production bug just before a deadline, how would you respond?",

    "If your application becomes slow after deployment, how would you investigate it?",

    "If you are given a technology you have never used before and a short deadline, what would you do?",

    "If your manager asks you to implement something you believe is technically incorrect, how would you handle it?"
]


def generate_situational_questions(count):

    questions = situational_questions.copy()

    random.shuffle(questions)

    result = []

    for question in questions[:count]:

        result.append({

            "question": question,

            "category": "Situational",

            "skill": "General"

        })

    return result


situational_count = question_distribution.get(
    "Situational",
    0
)

situational_generated = generate_situational_questions(
    situational_count
)
````

````python
# ==========================================================
# COMBINE ALL QUESTIONS
# ==========================================================

all_questions = (

    technical_generated

    + problem_generated

    + behavioral_generated

    + hr_generated

    + situational_generated

)


random.shuffle(all_questions)


# Add question IDs
for index, item in enumerate(all_questions, 1):

    item["question_id"] = index


print("=" * 60)
print("QUESTION GENERATION COMPLETED")
print("=" * 60)

print("Total Questions :", len(all_questions))

for item in all_questions:

    print(
        f"\nQ{item['question_id']}. "
        f"{item['question']}"
    )

    print("Category :", item["category"])
    print("Skill    :", item["skill"])
````

Now we'll build the Answer Evaluation Engine.

````python
# ==========================================================
# PHASE 6 : ANSWER EVALUATION
# ==========================================================

import json
import re
from datetime import datetime

with open(
    "../reports/interview_session.json",
    "r",
    encoding="utf-8"
) as file:

    interview_session = json.load(file)


print("=" * 60)
print("INTERVIEW SESSION LOADED")
print("=" * 60)

print("Role       :", interview_session["job_role"])
print("Questions  :", interview_session["total_questions"])
print("Answers    :", len(interview_session["answers"]))
print("Status     :", interview_session["status"])
````

````python
# ==========================================================
# BASIC ANSWER QUALITY
# ==========================================================

def calculate_answer_length_score(answer):

    words = answer.split()

    word_count = len(words)

    if word_count == 0:
        return 0

    elif word_count < 5:
        return 2

    elif word_count < 15:
        return 5

    elif word_count < 30:
        return 8

    else:
        return 10


print("=" * 60)
print("ANSWER LENGTH FUNCTION READY")
print("=" * 60)
````

````python
# ==========================================================
# KEYWORD EXTRACTION
# ==========================================================

stop_words = {
    "the",
    "is",
    "a",
    "an",
    "and",
    "or",
    "to",
    "of",
    "in",
    "on",
    "for",
    "with",
    "that",
    "this",
    "it",
    "are",
    "was",
    "were",
    "i",
    "we",
    "you",
    "they",
    "he",
    "she",
    "my",
    "our",
    "your",
    "be",
    "as",
    "at",
    "by",
    "from"
}


def extract_keywords(text):

    words = re.findall(
        r"\b[a-zA-Z][a-zA-Z0-9+#.-]*\b",
        text.lower()
    )

    keywords = []

    for word in words:

        if (
            word not in stop_words
            and len(word) > 2
        ):

            keywords.append(word)

    return set(keywords)


print("✓ Keyword extraction ready")
````

````python
# ==========================================================
# EXPECTED KEYWORDS
# ==========================================================

category_keywords = {

    "Technical": {
        "machine learning": [
            "model",
            "training",
            "testing",
            "data",
            "algorithm",
            "overfitting",
            "validation",
            "performance"
        ],

        "python": [
            "list",
            "tuple",
            "dictionary",
            "set",
            "mutable",
            "immutable"
        ],

        "general": [
            "project",
            "problem",
            "solution",
            "testing",
            "debugging"
        ]
    },


    "Problem Solving": [
        "algorithm",
        "complexity",
        "solution",
        "input",
        "output",
        "optimization",
        "time",
        "space"
    ],


    "Behavioral": [
        "situation",
        "problem",
        "action",
        "result",
        "team",
        "experience",
        "learned"
    ],


    "HR": [
        "experience",
        "skills",
        "strength",
        "career",
        "goal",
        "team",
        "company"
    ],


    "Situational": [
        "problem",
        "approach",
        "solution",
        "communication",
        "team",
        "testing",
        "result"
    ]
}


print("=" * 60)
print("EXPECTED KEYWORDS LOADED")
print("=" * 60)
````

````python
# ==========================================================
# GET EXPECTED KEYWORDS
# ==========================================================

def get_expected_keywords(
    category,
    skill
):

    if category == "Technical":

        skill_lower = skill.lower()

        if skill_lower in category_keywords[
            "Technical"
        ]:

            return category_keywords[
                "Technical"
            ][skill_lower]

        return category_keywords[
            "Technical"
        ]["general"]


    return category_keywords.get(
        category,
        []
    )


print(
    get_expected_keywords(
        "Problem Solving",
        "Problem Solving"
    )
)
````

````python
# ==========================================================
# KEYWORD COVERAGE
# ==========================================================

def calculate_keyword_score(
    answer,
    expected_keywords
):

    if not expected_keywords:

        return 5, []


    answer_words = extract_keywords(answer)

    matched_keywords = []


    for keyword in expected_keywords:

        keyword_parts = keyword.lower().split()

        if all(
            part in answer_words
            for part in keyword_parts
        ):

            matched_keywords.append(
                keyword
            )


    coverage = (
        len(matched_keywords)
        /
        len(expected_keywords)
    )


    score = round(
        coverage * 10,
        2
    )


    return score, matched_keywords
````

````python
# ==========================================================
# COMPLETENESS SCORE
# ==========================================================

def calculate_completeness_score(
    answer,
    question
):

    words = answer.split()

    if len(words) < 5:

        return 2

    if len(words) < 15:

        return 5

    if len(words) < 30:

        return 8

    return 10
````

````python
# ==========================================================
# EVALUATE ONE ANSWER
# ==========================================================

def evaluate_answer(answer_record):

    answer = answer_record["answer"]

    category = answer_record["category"]

    skill = answer_record["skill"]

    question = answer_record["question"]


    # ------------------------------------------
    # 1. Length Score
    # ------------------------------------------

    length_score = calculate_answer_length_score(
        answer
    )


    # ------------------------------------------
    # 2. Expected Keywords
    # ------------------------------------------

    expected_keywords = get_expected_keywords(
        category,
        skill
    )


    # ------------------------------------------
    # 3. Keyword Score
    # ------------------------------------------

    keyword_score, matched_keywords = (
        calculate_keyword_score(
            answer,
            expected_keywords
        )
    )


    # ------------------------------------------
    # 4. Completeness
    # ------------------------------------------

    completeness_score = (
        calculate_completeness_score(
            answer,
            question
        )
    )


    # ------------------------------------------
    # 5. Overall Score
    # ------------------------------------------

    overall_score = round(

        (
            length_score * 0.20
            +
            keyword_score * 0.50
            +
            completeness_score * 0.30

        ),

        2
    )


    return {

        "question_id":
            answer_record["question_id"],

        "question":
            question,

        "category":
            category,

        "skill":
            skill,

        "answer":
            answer,

        "length_score":
            length_score,

        "keyword_score":
            keyword_score,

        "completeness_score":
            completeness_score,

        "matched_keywords":
            matched_keywords,

        "overall_score":
            overall_score

    }
````

````python
# ============================================================
# PHASE 7 - AI FEEDBACK
# Cell 3: Gemini Feedback Engine
# ============================================================

# ------------------------------------------------------------
# Initialize Gemini
# ------------------------------------------------------------
# ...existing code...
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

MODEL_NAME = "gemini-3.5-flash"
FALLBACK_MODELS = ["chat-bison-001", "text-bison-001"]

def init_llm(model_name):
    return ChatGoogleGenerativeAI(
        model=model_name,
        google_api_key=GOOGLE_API_KEY,
        temperature=0.3,
    )

llm = None
last_err = None
for m in [MODEL_NAME] + FALLBACK_MODELS:
    try:
        llm = init_llm(m)
        print(f"Initialized GenAI model: {m}")
        break
    except Exception as e:
        last_err = e
        print(f"Model init failed for {m}: {e}")
        # try next fallback if model not found
        continue

if llm is None:
    raise RuntimeError(
        "Could not initialize any GenAI model. "
        "Check GOOGLE_API_KEY and available models. Last error: "
        f"{last_err}"
    )
# ...existing code...

# ------------------------------------------------------------
# Feedback Prompt
# ------------------------------------------------------------

# ...existing code...
feedback_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            """
You are an expert AI interview coach.

You are given an interview evaluation report generated by
another evaluation system.

Your task is to analyze the candidate's performance and provide
constructive, personalized interview feedback.

Do NOT invent candidate information.

Base your feedback ONLY on the supplied evaluation report.

Return ONLY valid JSON.

The JSON must follow this structure:

{
    "overall_feedback": "string",

    "overall_assessment": {
        "performance_level": "Excellent | Good | Average | Needs Improvement",
        "summary": "string"
    },

    "strengths": [
        "string"
    ],

    "weaknesses": [
        "string"
    ],

    "technical_feedback": [
        "string"
    ],

    "communication_feedback": [
        "string"
    ],

    "improvement_areas": [
        {
            "area": "string",
            "priority": "High | Medium | Low",
            "reason": "string"
        }
    ],

    "recommended_topics": [
        {
            "topic": "string",
            "reason": "string"
        }
    ],

    "study_plan": [
        {
            "day": 1,
            "focus": "string",
            "action": "string"
        }
    ],

    "interview_advice": [
        "string"
    ]
}

Rules:

1. Do not invent scores.
2. Do not invent technologies.
3. Do not invent questions.
4. Use evidence from the evaluation report.
5. Be constructive.
6. Give actionable recommendations.
7. Keep the study plan practical.
8. Return JSON only.
"""
        ),
        (
            "human",
            """
Here is the interview evaluation report:

{{ evaluation_report }}

Generate the personalized AI feedback report.
"""
        ),
    ],
    template_format="jinja2",
)
# ...existing code...

# ------------------------------------------------------------
# Convert evaluation report to text
# ------------------------------------------------------------

evaluation_text = json.dumps(
    evaluation_data,
    indent=4,
    ensure_ascii=False
)


# ------------------------------------------------------------
# Create prompt
# ------------------------------------------------------------

messages = feedback_prompt.format_messages(
    evaluation_report=evaluation_text
)

print("=" * 70)
print("GENERATING AI FEEDBACK...")
print("=" * 70)


# ------------------------------------------------------------
# Call Gemini
# ------------------------------------------------------------

try:

    response = llm.invoke(messages)

    raw_feedback = response.content

    print("AI feedback generated successfully.")

except Exception as e:

    print("=" * 70)
    print("LLM ERROR")
    print("=" * 70)

    print(type(e).__name__)
    print(str(e))

    raise
````

````python
# ============================================================
# PHASE 7 - AI FEEDBACK
# Cell 4: Robust JSON Parsing + Validation + Save
# ============================================================

import json
import re


def normalize_llm_response(response_content):
    """
    Convert different Gemini/LangChain response formats
    into a plain string.
    """

    # --------------------------------------------------------
    # Case 1: Already a string
    # --------------------------------------------------------

    if isinstance(response_content, str):
        return response_content.strip()

    # --------------------------------------------------------
    # Case 2: List response
    # --------------------------------------------------------

    if isinstance(response_content, list):

        text_parts = []

        for item in response_content:

            # ----------------------------------------------
            # Dictionary format
            # ----------------------------------------------

            if isinstance(item, dict):

                if "text" in item:
                    text_parts.append(
                        str(item["text"])
                    )

                elif item.get("type") == "text":
                    text_parts.append(
                        str(item.get("text", ""))
                    )

            # ----------------------------------------------
            # Plain string inside list
            # ----------------------------------------------

            elif isinstance(item, str):

                text_parts.append(item)

        return "\n".join(text_parts).strip()

    # --------------------------------------------------------
    # Case 3: Object with text attribute
    # --------------------------------------------------------

    if hasattr(response_content, "text"):

        return str(
            response_content.text
        ).strip()

    # --------------------------------------------------------
    # Fallback
    # --------------------------------------------------------

    return str(response_content).strip()


def extract_json(text):
    """
    Extract JSON from Gemini response.

    Handles:

    1. Plain JSON
    2. ```json ... ```
    3. JSON surrounded by explanation
    """

    # --------------------------------------------------------
    # Normalize
    # --------------------------------------------------------

    text = normalize_llm_response(text)

    if not text:

        raise ValueError(
            "LLM returned an empty response."
        )

    # --------------------------------------------------------
    # Remove Markdown code fences
    # --------------------------------------------------------

    text = re.sub(
        r"```json\s*",
        "",
        text,
        flags=re.IGNORECASE
    )

    text = re.sub(
        r"```\s*",
        "",
        text
    )

    text = text.strip()

    # --------------------------------------------------------
    # Attempt 1: Entire response is JSON
    # --------------------------------------------------------

    try:

        return json.loads(text)

    except json.JSONDecodeError:
        pass

    # --------------------------------------------------------
    # Attempt 2: Extract JSON object
    # --------------------------------------------------------

    start = text.find("{")
    end = text.rfind("}")

    if start != -1 and end != -1:

        json_text = text[
            start:end + 1
        ]

        try:

            return json.loads(
                json_text
            )

        except json.JSONDecodeError:
            pass

    # --------------------------------------------------------
    # Attempt 3: Extract JSON array
    # --------------------------------------------------------

    start = text.find("[")
    end = text.rfind("]")

    if start != -1 and end != -1:

        json_text = text[
            start:end + 1
        ]

        try:

            return json.loads(
                json_text
            )

        except json.JSONDecodeError:
            pass

    # --------------------------------------------------------
    # Failed
    # --------------------------------------------------------

    print("=" * 70)
    print("RAW LLM RESPONSE")
    print("=" * 70)

    print(text)

    print("=" * 70)

    raise ValueError(
        "Gemini did not return valid JSON."
    )


# ============================================================
# Parse Gemini Response
# ============================================================

try:

    feedback_data = extract_json(
        raw_feedback
    )

except Exception as e:

    print("=" * 70)
    print("JSON PARSING ERROR")
    print("=" * 70)

    print(
        f"Error type: {type(e).__name__}"
    )

    print(
        f"Error: {e}"
    )

    raise


# ============================================================
# Validate Output
# ============================================================

required_fields = [

    "overall_feedback",

    "overall_assessment",

    "strengths",

    "weaknesses",

    "technical_feedback",

    "communication_feedback",

    "improvement_areas",

    "recommended_topics",

    "study_plan",

    "interview_advice"

]


missing_fields = [

    field

    for field in required_fields

    if field not in feedback_data

]


if missing_fields:

    print("=" * 70)
    print("INVALID FEEDBACK STRUCTURE")
    print("=" * 70)

    print(
        "Missing fields:"
    )

    for field in missing_fields:

        print(
            f" - {field}"
        )

    raise ValueError(
        "AI feedback is missing required fields."
    )


# ============================================================
# Save Feedback
# ============================================================

with open(
    FEEDBACK_FILE,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        feedback_data,
        file,
        indent=4,
        ensure_ascii=False
    )


# ============================================================
# Display Result
# ============================================================

print("=" * 70)
print("PHASE 7 COMPLETED SUCCESSFULLY")
print("=" * 70)

print(
    f"Feedback saved to:\n{FEEDBACK_FILE}"
)

print("=" * 70)

print(
    json.dumps(
        feedback_data,
        indent=4,
        ensure_ascii=False
    )
)
````
